`time_gap_minutes`: minutes since the previous record for the same person, used to detect long pauses/rests.

`prev_location`: previous area for the same person.

`location_changed`: True only when the person changed from one area to another area. This is the main signal for real movement/walking.

`is_gap`: True if `time_gap_minutes` > 15, used to detect a rest/inactive gap. This is not the same as walking.

`episode_id`: groups continuous tracking records per person. It increments after a long gap.

`movement_status`: high-level label: `moved`, `rested`, or `continuous`.

`movement_detail`: detailed label: `session_start`, `moved`, `moved_after_gap`, `rested`, or `stationary`.

```python
df['timestamp'] = pd.to_datetime(df['timestamp']).dt.floor('s').dt.tz_localize(None)

# Movement must be calculated per person in timestamp order.
df = df.sort_values(['personId', 'timestamp']).reset_index(drop=True)

# Previous record context for the same tracked person.
df['prev_timestamp'] = df.groupby('personId')['timestamp'].shift()
df['prev_location'] = df.groupby('personId')['area'].shift()

# Minutes since this person's previous location update.
df['time_gap_minutes'] = (df['timestamp'] - df['prev_timestamp']).dt.total_seconds() / 60

# True movement means the person changed area compared with their previous record.
df['location_changed'] = (
    df['prev_location'].notna()
    & df['area'].notna()
    & (df['area'] != df['prev_location'])
)

# A gap/rest is a long pause between updates. It is not automatically movement.
df['is_gap'] = df['time_gap_minutes'].gt(15).fillna(False)

# Continuous tracking episode per person. A new episode starts after a long gap.
df['episode_id'] = df.groupby('personId')['is_gap'].cumsum().astype(int)

# Running count of real area changes inside each episode.
df['location_in_episode'] = (
    df.groupby(['personId', 'episode_id'])['location_changed']
    .cumsum()
    .astype(int)
)

# High-level status for metric filtering.
df['movement_status'] = np.select(
    [df['location_changed'], df['is_gap']],
    ['moved', 'rested'],
    default='continuous',
)

# Detailed label for table answers and future exercise/walking questions.
df['movement_detail'] = np.select(
    [
        df['prev_location'].isna(),
        df['location_changed'] & df['is_gap'],
        df['location_changed'],
        df['is_gap'],
    ],
    [
        'session_start',
        'moved_after_gap',
        'moved',
        'rested',
    ],
    default='stationary',
)

# Drop helper column; keep the fact table schema stable.
df = df.drop(columns=['prev_timestamp'])

# Keep the final fact table easy to inspect chronologically.
df = df.sort_values(['timestamp', 'personId']).reset_index(drop=True)

print(df[['personFirstName', 'timestamp', 'area', 'prev_location', 'time_gap_minutes', 'location_changed', 'is_gap', 'episode_id', 'movement_status', 'movement_detail']])
```

`is_gap` - True if `time_gap_minutes` > 15, signals a pause/rest between movement sessions.

`episode_id` - counter that increments each time gap > 15 min is detected. It groups one continuous tracking session per person.

`movement_status` - high-level movement label: `moved` when area changed, `rested` when there was a long gap, otherwise `continuous`.

`location_changed` - `True` if the patient/person is in a different `area` than their previous record. This is the main signal for walking/movement.

`location_in_episode` - running count of how many real area changes occurred within the current episode.

`movement_detail` - detailed movement label: `session_start`, `moved_after_gap`, `moved`, `rested`, or `stationary`.

`prev_location` - name of the area the patient/person was in before the current record. `NaN` on the first record per person.

- `location_changed` + `prev_location` -> Track where a person came from and whether they really moved.
- `movement_status = moved` -> Count true movement events.
- `time_gap_minutes` + `is_gap` -> Detect resting/inactive periods.
- `episode_id` + `location_in_episode` -> Group movement activity into sessions.
